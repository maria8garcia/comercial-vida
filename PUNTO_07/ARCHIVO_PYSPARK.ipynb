```python
#IMPORTANDO LIBRERIAS - Leer archivos, Crear dataframes y consultas con spark sql
from pyspark.sql import SparkSession  
from pyspark.sql.functions import sum, avg, count, min, max, round
spark = SparkSession.builder \
    .appName("AnalisisVentas") \
    .getOrCreate()

print("Spark iniciado correctamente")
```

### Carga del archivo CSV

Se carga el archivo ventas.csv utilizando PySpark.
Se utiliza `header=True` para reconocer los nombres de las columnas e
`inferSchema=True` para que Spark detecte automáticamente los tipos de datos.

```python
#LECTURA DE ARCHIVO CSV
df_ventas = spark.read.csv(
    "ventas.csv",
    header=True,
    inferSchema=True
)

df_ventas.show()
```

### Revisión de la estructura del DataFrame

Se revisa el esquema de los datos para identificar los tipos de datos asignados
por PySpark a cada columna.

```python
#VER ESTRUCTURA DEL CSV
df_ventas.printSchema()
```

### Consulta utilizando DataFrames

Se seleccionan las columnas principales de las ventas mediante la API
de DataFrames de PySpark.

```python
#COLUMNAS PRINCIPALES DEL DATAFRAME
df_ventas.select(
    "IdVenta",
    "Fecha",
    "IdCliente",
    "TotalVenta"
).show()
```

### Uso de SparkSQL

Para realizar consultas SQL sobre el DataFrame se crea una vista temporal
denominada `ventas`.

```python
#REGISTRO TEMPORAL PARA CONSULTAS SQL
df_ventas.createOrReplaceTempView("ventas")
```

```python
resultado_sql = spark.sql("""
    SELECT
        IdVenta,
        Fecha,
        IdCliente,
        TotalVenta
    FROM ventas
""")

resultado_sql.show()
```

## 7.2 Agregación y estadística descriptiva

Se realizaron operaciones de agregación sobre los datos de ventas con el objetivo
de obtener indicadores generales como la cantidad de ventas, el total vendido,
el promedio de venta, el valor mínimo y el valor máximo.

```python
from pyspark.sql.functions import sum, avg, count, min, max, round
```

```python
resultado_agregacion = df_ventas.agg(
    count("IdVenta").alias("CantidadVentas"),
    round(sum("TotalVenta"), 2).alias("TotalVendido"),
    round(avg("TotalVenta"), 2).alias("PromedioVenta"),
    min("TotalVenta").alias("VentaMinima"),
    max("TotalVenta").alias("VentaMaxima")
)
resultado_agregacion.show()
```

La agregación permitió resumir los datos de ventas mediante indicadores generales.
Se calculó la cantidad total de ventas registradas, el importe acumulado, el
promedio de venta y los valores mínimo y máximo.

Estos indicadores permiten analizar el comportamiento general de las ventas sin necesidad de revisar individualmente cada registro.

## ESTADÍSTICA DESCRIPTIVA
Mediante la función `describe()` se obtuvieron estadísticas descriptivas de la
columna TotalVenta, incluyendo cantidad de registros, promedio, desviación
estándar, valor mínimo y valor máximo.

```python
df_ventas.describe("TotalVenta").show()
```

## CON SPARKSQL
Trabaja de la misma manera que DATAFRAME solo que la sintaxis
es muy parecida a SQL

```python
spark.sql("""
    SELECT
        COUNT(IdVenta) AS CantidadVentas,
        ROUND(SUM(TotalVenta), 2) AS TotalVendido,
        ROUND(AVG(TotalVenta), 2) AS PromedioVenta,
        MIN(TotalVenta) AS VentaMinima,
        MAX(TotalVenta) AS VentaMaxima
    FROM ventas
""").show()
```
